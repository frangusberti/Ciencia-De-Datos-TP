# Exploración del Dataset Limpio
Este archivo te permite visualizar el dataset `customer_behavior_LIMPIO.csv` en formato tabla, con las columnas traducidas al español, ideal para empezar tu fase de modelo predictivo.

```python
import pandas as pd

# Permitir que Pandas muestre todas las columnas
pd.set_option('display.max_columns', None)

# Cargar el dataset limpio
df = pd.read_csv('customer_behavior_LIMPIO.csv')

# Renombrar las columnas basándonos en descripcion_de_columnas.txt
nuevos_nombres = {
    'Education': 'Educacion',
    'Marital_Status': 'Estado_Civil',
    'MntWines': 'Gasto_Vinos',
    'MntFruits': 'Gasto_Frutas',
    'MntMeatProducts': 'Gasto_Carnes',
    'MntFishProducts': 'Gasto_Pescados',
    'MntSweetProducts': 'Gasto_Dulces',
    'MntGoldProds': 'Gasto_Premium',
    'NumDealsPurchases': 'Compras_Con_Descuento',
    'NumWebPurchases': 'Compras_Web',
    'NumCatalogPurchases': 'Compras_Catalogo',
    'NumStorePurchases': 'Compras_Tienda',
    'NumWebVisitsMonth': 'Visitas_Web_Mes',
    'AcceptedCmp1': 'Acepto_Cmp1',
    'AcceptedCmp2': 'Acepto_Cmp2',
    'AcceptedCmp3': 'Acepto_Cmp3',
    'AcceptedCmp4': 'Acepto_Cmp4',
    'AcceptedCmp5': 'Acepto_Cmp5',
    'Complain': 'Queja',
    'Response': 'Respuesta_Ultima_Cmp'
}

df = df.rename(columns=nuevos_nombres)

# Visualizar las primeras 20 filas en formato tabla interactiva
df.head(20)
```
